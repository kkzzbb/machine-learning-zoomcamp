```python
asia_mask = df['origin'].astype(str).str.strip().str.lower() == 'asia'
asia_cars = df[asia_mask]

asia_cars['fuel_efficiency_mpg'].max()
```

```python
median_before = df['horsepower'].median()
mode_value = df['horsepower'].mode()[0]
print(f"Median before: {median_before}")
print(f"Mode (most frequent): {mode_value}")
```

```python
df_filled = df.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(mode_value)
median_after = df_filled['horsepower'].median()
print(f"Median after: {median_after}")
if median_after > median_before:
    print("Yes, it increased")
elif median_after < median_before:
    print("Yes, it decreased")
else:
    print("No")
```

```python
selected_cols = asia_cars[['vehicle_weight','model_year']]
first_7 = selected_cols.iloc[:7]
X = first_7.values
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv.dot(X.T).dot(y) 
```

```python
print(f"w: {w}")
print(f"Sum of elements: {w.sum():.4f}")
```
